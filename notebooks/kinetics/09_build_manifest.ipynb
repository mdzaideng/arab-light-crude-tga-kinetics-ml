# 09_build_manifest

Notebook conversion of `09_build_manifest.py`.

- Original computational/output logic is preserved.
- `__file__`-dependent paths are adapted for Jupyter execution.
- This script does not create figures, so no font configuration is required.


```python
from __future__ import annotations
from pathlib import Path
import csv, hashlib
# Notebook-safe package root
CODE_DIR = Path.cwd().resolve()
if (CODE_DIR / "03_CODE").exists():
    CODE_DIR = CODE_DIR / "03_CODE"
ROOT = CODE_DIR.parent
OUT = ROOT / "00_PACKAGE_CONTROL" / "MANIFEST_SHA256.csv"
rows=[]
for p in sorted(ROOT.rglob("*")):
    if p.is_file() and p != OUT and "__pycache__" not in p.parts:
        rows.append((str(p.relative_to(ROOT)), p.stat().st_size, hashlib.sha256(p.read_bytes()).hexdigest()))
OUT.parent.mkdir(parents=True, exist_ok=True)
with OUT.open("w", newline="", encoding="utf-8") as f:
    w=csv.writer(f); w.writerow(["relative_path","bytes","sha256"]); w.writerows(rows)
print(f"Wrote {OUT} with {len(rows)} entries")
```
